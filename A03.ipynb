# Cambiar variables categóricas a numéricas

```python
df["Tamaño"] = df["Tamaño"].replace({"Chica": 0, "Grande": 1})
df["Hielo"] = df["Hielo"].replace({"Si": 1, "No": 0})
```

```python
df.head()
```

```python
df["Naranja"] = df["Sabor"].str.contains("Naranja")
df["Coco"] = df["Sabor"].str.contains("Coco")
df["Kahlua"] = df["Sabor"].str.contains("Kahlua")
df["Fresa-piña"] = df["Sabor"].str.contains("Fresa-piña")
df["Sandia"] = df["Sabor"].str.contains("Sandia")
```

```python
df.head()
```

```python
df = df.drop("Sabor", axis=1)
```

```python
df.head()
```

# Regresión lineal

```python
import statsmodels.api as sm
```

```python
x = df[['Tamaño', 'Hielo', 'Naranja', 'Coco', 'Kahlua', 'Fresa-piña', 'Sandia']].astype(int)
x = sm.add_constant(x)
y = df['Calificación']
```

```python
ols = sm.OLS(y, x)
results = ols.fit()
results.summary()
```

```python
tam_encoder = OrdinalEncoder()
tam_encoder.fit(df[["Tamaño"]])
df["Tamaño"] = tam_encoder.transform(df[["Tamaño"]])
```

```python
hie_encoder = OrdinalEncoder()
hie_encoder.fit(df[["Hielo"]])
df["Hielo"] = hie_encoder.transform(df[["Hielo"]])
```

```python
sab_encoder = OneHotEncoder(categories=[["Naranja", "Coco", "Kahlua", "Fresa-piña", "Sandia"]], sparse_output=False)
sab_encoder.fit(df[["Sabor"]])
df["Sabor"] = sab_encoder.transform(df[["Sabor"]])
```

# Verificamos con regresión lineal

```python
x = df[["Sabor", "Tamaño", "Hielo"]]
x = sm.add_constant(x)
y = df['Calificación']
```

```python
ols = sm.OLS(y, x)
results = ols.fit()
results.summary()
```
