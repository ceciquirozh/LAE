```python
import pandas as pd
import matplotlib.pyplot as plt
from sklearn.linear_model import LinearRegression
import numpy as np
```

```python
df = pd.read_csv("Salarios_minimos.csv")
```

```python
df.describe()
```

```python
x = df["Periodo"]
y = df["Salario"]
```

```python
lr = LinearRegression()
```

```python
x_v = np.reshape(x,[-1,1])
x_v
```

```python
X = x_v
```

```python
lr.fit(X,y)
```

```python
lr.intercept_
```

```python
lr.coef_
```

```python
def f1(x): return lr.intercept_ + lr.coef_*x
```

```python
plt.figure()
plt.scatter(x, f1(x))
plt.scatter(x, y)
```
